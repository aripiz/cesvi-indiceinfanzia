# Cesvi Indice Infanzia - Esplorazione dati

```python
import pandas as pd
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("cesvi-indiceinfanzia_data.csv")
```

```python
df
```

```python
df.pivot_table(index="territory", values="score", columns="year", aggfunc="mean")
```

```python
# Prepariamo i dati per la heatmap (Indicatori vs Regioni nel 2024)
data_2024 = df[df['year'] == 2024].pivot(index='territory', columns='indicator', values='score')

plt.figure(figsize=(12, 8))
sns.heatmap(data_2024, cmap='RdYlGn', annot=False) # Rosso=Basso, Verde=Alto
plt.title('Performance Regionali per Indicatore (2024)')
plt.show()
```
